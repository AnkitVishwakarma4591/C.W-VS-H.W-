```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Number of rows
num_rows = 300

# Generate independent variables (features)
X1 = np.random.uniform(-10, 10, num_rows)  # Feature 1
X2 = np.random.uniform(-5, 5, num_rows)    # Feature 2
X3 = np.random.uniform(0, 15, num_rows)    # Feature 3

# Generate polynomial terms (e.g., squared and interaction terms)
X1_squared = X1 ** 2
X2_squared = X2 ** 2
X1_X2 = X1 * X2

# Generate a logistic function for probability
linear_combination = 0.5 * X1 + 0.3 * X2 + 0.2 * X3 - 0.1 * X1_squared + 0.05 * X2_squared + 0.1 * X1_X2
probabilities = 1 / (1 + np.exp(-linear_combination))  # Sigmoid function

# Convert probabilities to binary class labels (0 or 1) based on a threshold of 0.5
y = (probabilities > 0.5).astype(int)

# Create DataFrame
df = pd.DataFrame({
    "X1": X1,
    "X2": X2,
    "X3": X3,
    "X1_squared": X1_squared,
    "X2_squared": X2_squared,
    "X1_X2": X1_X2,
    "Output": y
})

# Save to CSV file
df.to_csv("polynomial_logistic_regression_data.csv", index=False)

# Return the file path
csv_filename
```
